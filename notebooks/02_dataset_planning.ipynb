```python
import pandas as pd
pumps = pd.read_csv("../pump_telegram.csv")
pumps.head()
```

```python
binance_pumps = pumps[
    pumps["exchange"].str.lower() == "binance"
].copy()

binance_pumps.shape
```

```python
binance_pumps["pair"] =  (
    binance_pumps["symbol"].astype(str) + "BTC"
)
binance_pumps.head()
```

```python
binance_pumps["date"] = pd.to_datetime(
    binance_pumps["date"]
)
binance_pumps["year"] = binance_pumps["date"].dt.year
binance_pumps["month"] = binance_pumps["date"].dt.month
binance_pumps[["symbol", "pair", "date", "hour", "year", "month"]].head(20)
```
